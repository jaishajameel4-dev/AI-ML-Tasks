## Load and Inspect the Iris Dataset

First, we'll load the Iris dataset using pandas, which can be easily accessed through the seaborn library. Then, we'll inspect its shape, column names, and the first few rows.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the Iris dataset
iris_df = sns.load_dataset('iris')

print("Dataset loaded successfully!")
```

### Dataset Shape, Column Names, and Head

```python
# Print the shape of the dataset
print(f"Shape of the dataset: {iris_df.shape}")

# Print the column names
print("\nColumn names:")
for col in iris_df.columns:
    print(f"- {col}")

# Display the first few rows of the dataset
print("\nFirst 5 rows of the dataset:")
display(iris_df.head())
```

## Summary Statistics

Now, let's use `.info()` and `.describe()` to get a quick overview of the data types, non-null values, and statistical summary of the numerical columns.

```python
# Display concise summary of the DataFrame
print("\nDataFrame Information:")
iris_df.info()

# Display descriptive statistics
print("\nDescriptive Statistics:")
display(iris_df.describe())
```

## Data Visualization

Next, we'll visualize the dataset to understand relationships between features, value distributions, and identify potential outliers.

### Scatter Plot: Relationships between Features

A pair plot is an excellent way to visualize the pairwise relationships between all features in the dataset, colored by species to observe separation.

```python
# Create a scatter plot to show relationships between features
sns.pairplot(iris_df, hue='species', diag_kind='kde')
plt.suptitle('Pair Plot of Iris Dataset Features', y=1.02) # Add a title above the plots
plt.show()
```

### Histograms: Value Distributions

Histograms help us understand the distribution of each numerical feature.

```python
# Create histograms for all numerical columns
iris_df.hist(figsize=(10, 8))
plt.suptitle('Histograms of Iris Dataset Features', y=1.02) # Add a title above the plots
plt.tight_layout(rect=[0, 0.03, 1, 0.98]) # Adjust layout to prevent title overlap
plt.show()
```

### Box Plots: Identifying Outliers

Box plots are useful for visualizing the distribution of numerical data and identifying potential outliers for each feature, broken down by species.

```python
# Create box plots for each numerical feature, colored by species
plt.figure(figsize=(12, 10))
for i, feature in enumerate(iris_df.columns[:-1]): # Exclude the 'species' column
    plt.subplot(2, 2, i + 1) # 2 rows, 2 columns for 4 features
    sns.boxplot(x='species', y=feature, data=iris_df)
    plt.title(f'Box Plot of {feature} by Species')
plt.tight_layout()
plt.suptitle('Box Plots of Iris Dataset Features by Species', y=1.02) # Add a title above all subplots
plt.show()
```
