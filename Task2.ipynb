```python
import yfinance as yf
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Select a stock
stock_ticker = 'AAPL' # Example: Apple Inc.

# 2. Load historical data
data = yf.download(stock_ticker, start='2020-01-01', end='2023-01-01')

print(f"Downloaded {len(data)} rows of data for {stock_ticker}.")
print(data.head())

# 3. Feature Engineering: Use Open, High, Low, Volume to predict next Close price
# Create the target variable: next day's Close price
data['Next_Close'] = data['Close'].shift(-1)

# Drop the last row as it won't have a 'Next_Close' value
data.dropna(inplace=True)

# Define features (X) and target (y)
features = ['Open', 'High', 'Low', 'Volume', 'Close'] # Including 'Close' as a feature for current day
X = data[features]
y = data['Next_Close']

print("\nFeatures (X) head:")
print(X.head())
print("\nTarget (y) head:")
print(y.head())
```
